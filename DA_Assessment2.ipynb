```python
# Write a function that takes the radius of a circle and returns its area.
```

```python
import math
def circle_area(radius):
    return math.pi * radius * radius

r = float(input("Enter radius: "))
print("Area:", circle_area(r))
```

```python
# Write a function that accepts three numbers and returns the largest one.
```

```python

def max_num(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(max_num(20, 40, 30))
```

```python
#Write a function that takes a string and returns its length without using built-in length functions.
```

```python
def string_length(s):
    count = 0
    for ch in s:
        count += 1
    return count

print(string_length("Data Analytics"))
```

```python
# Function to check whether a number is prime
```

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(7))
print(is_prime(10))
```

```python
# Write a recursive function to find the nth Fibonacci number.
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(8))
```

```python
# Write a recursive function to reverse a string.
```

```python
def reverse_string(s):
    rev = ""
    for ch in s:
        rev = ch + rev
    return rev

print(reverse_string("Python"))
```

```python
# Write a program to find the sum of all numbers from 1 to n using a for loop.
```

```python
def sum(a):
    total=0
    while(a>0):
        total=total+a
        a=a-1
    return total
print(sum(5))
```
